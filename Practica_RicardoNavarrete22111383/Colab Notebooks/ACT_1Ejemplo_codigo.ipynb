En esta actividad usarás el archivo de datos `Forbes_2000_top_company.csv`. Este conjunto contiene las 2000 principales empresas, clasificadas mediante el uso de cuatro métricas: ventas, ganancias, activos y valor de mercado. Tiene las 11 columnas de datos:

*   `2022 Ranking`: Ranking del año actual de la empresa
*   `Organization Name`: Nombre
*   `Industry`: El tipo de industria con la que trata principalmente
*   `Country`: País de origen
*   `Year Founded`: Año en que se fundó
*   `CEO`: CEO
*   `Profits (Billions)`: Ganancias del año
*   `Assets (Billions)`: Activos del año
*   `Market Value (Billions)`: Valor de mercado del año
*   `Total Employees`: Total de empleados

https://www.kaggle.com/datasets/ash316/forbes-top-2000-companies

```python
import matplotlib.pyplot as plt;
 [forbes_df[col].value_counts().head(15).plot(kind='bar', figsize=(15, 5), title=col).get_figure().show() or plt.show() for col in forbes_df.select_dtypes(include=['object']).columns]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

#filtrar las columnas numéricas de forma automática
columnas_numericas = forbes_df.select_dtypes(include=['number']).columns

#el bucle genera cada gráfica con el tamaño y ancho perfecto
for col in columnas_numericas:
    #lienzo ancho e independiente para cada columna
    plt.figure(figsize=(20, 5))

    #boxplot y configurar la escala logarítmica para evitar que se junten
    ax = sns.boxplot(x=forbes_df[col], width=0.4, color="skyblue")
    ax.set_xscale('log')

    #colocar el nombre de la columna únicamente arriba como título
    ax.set_title(col, fontsize=12, fontweight='bold', pad=10)
    ax.set_xlabel('') #limpiar textos basura

    #renderizado de pantalla
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

#ventas y ganancias
plt.figure(figsize=(14, 6))

#alpha=0.6 para ver si hay puntos encimados
sns.scatterplot(
    data=forbes_df,
    x='Revenue (Billions)',
    y='Profits (Billions)',
    alpha=0.6,
    color='purple',
    s=30 #puntos
)

#escala logaritmica para que no es junten mucho los datos
plt.xscale('log')
plt.yscale('log')

plt.title('Gráfica de Dispersión: Ventas vs Ganancias', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Ventas (Revenue in Billions)')
plt.ylabel('Ganancias (Profits in Billions)')
plt.show()


#activos y valor de mercado
plt.figure(figsize=(14, 6))

sns.scatterplot(
    data=forbes_df,
    x='Assets (Billions)',
    y='Market Value (Billions)',
    alpha=0.6,
    color='teal',
    s=60
)

plt.xscale('log')
plt.yscale('log')

plt.title('Gráfica de Dispersión: Activos vs Valor de Mercado', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Activos totales (Assets in Billions)')
plt.ylabel('Valor de Mercado (Market Value in Billions)')
plt.show()
```
